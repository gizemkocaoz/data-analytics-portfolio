# Stack Overflow Developer Survey Analysis

This project analyzes the Stack Overflow Developer Survey dataset using Pandas.

## 1. Import Libraries

```python
import pandas as pd
```

## 2. Load Dataset

```python
df = pd.read_csv(
    "Downloads/results.csv",
    low_memory=False
)
```

## 3. Explore Dataset

```python
df.shape
```

```python
df.info()
```

```python
df.isna().sum().sort_values(ascending=False).head(10)
```

## Question 1: How Many People Participated in the Survey?

```python
print("Number of survey participants:", df.shape[0])
```

## Question 2: How Many People Answered All Questions?

```python
answered_all = df.dropna()

answered_all.shape[0]
```

```python
df.notna().all(axis=1).sum()
```

**Note:** No participant answered every question in the survey. This is expected because many survey questions were optional or only shown to specific groups of respondents.

## Question 3: What are the measures of central tendency for WorkExp?

```python
pd.DataFrame({
    "Measure": ["Mean", "Median", "Mode"],
    "Value": [
        df["WorkExp"].mean(),
        df["WorkExp"].median(),
        df["WorkExp"].mode()[0]
    ]
})
```

## Question 4: How Many People Work Remotely?

```python
remote_workers = df[df["RemoteWork"] == "Remote"]

print("Number of remote workers:", remote_workers.shape[0])
```

## Question 5: What Percentage of Respondents Program in Python?

```python
python_users = df["LanguageHaveWorkedWith"].str.contains(
    "Python",
    na=False
)

python_percentage = (
    python_users.sum() / df.shape[0]
) * 100

print(f"Percentage of Python developers: {python_percentage:.2f}%")
```

## Question 6: How Many People Learned Programming Through Online Courses?

```python
online_course_users = df["LearnCode"].str.contains(
    "Online Courses",
    na=False
)

print("People who learned through online courses:",
      online_course_users.sum())
```

## Question 7: Average and Median Annual Compensation of Python Developers by Country

```python
salary_by_country = (
    df[df["LanguageHaveWorkedWith"].str.contains("Python", na=False)]
    .groupby("Country")["ConvertedCompYearly"]
    .agg(["mean", "median"])
    .sort_values("mean", ascending=False)
)

salary_by_country.head(20)
```

## Question 8: Education Levels of the Top 5 Highest-Paid Respondents

```python
top_5_education = (
    df.sort_values("ConvertedCompYearly", ascending=False)
    [["ConvertedCompYearly", "EdLevel"]]
    .head(5)
)

top_5_education
```
